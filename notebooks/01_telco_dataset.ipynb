O que entendemos do nosso primeiro contato com os dados:

- `costumerID` será o identificador único daquele usuário.

- Colunas categóricas serão transformadas por **OneHotEncoding**. Dessa forma, previnimos o nosso modelo de interpretar as representações numéricas das categorias de forma hierárquica.

- Colunas numéricas deverão ser normalizadas via **z-score**. A colunas `TotalCharges` é do tipo object, mas deverá passar para o tipo float64.

---
> Obs: `SeniorCitizen` é numérica mas representa um dado binário e, portanto, não será normalizada.

```python
df.nunique()
```

1. Colunas que passarão por **OneHotEnconding**:
   - ['gender', 'SeniorCitizen', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod', 'Churn'].

2. Colunas que passarão por normalização **z-score**:
   - ['tenure', 'MonthlyCharges', 'TotalCharges'].

---
Vamos agora, observar quais são as categorias de cada uma das colunas que irá passar pelo **OneHotEncoding**.

```python
AMARELO = '\033[93m'
RESET = '\033[0m'

onehot_columns = ['gender', 'SeniorCitizen', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod', 'Churn']

for col in onehot_columns:
    print(f"{AMARELO}{col}: {RESET}{df[col].unique()}")
```
